## Abstract

Housing affordability is one of the most pressing economic concerns facing American households, yet the extent to which local income levels actually explain variation in home prices across regions remains debated. In this study we ask whether median household income is associated with median home prices across U.S. metropolitan areas after controlling for local unemployment rates. Using 2023 data from the U.S. Census Bureau's American Community Survey (median household income, median home value) and the Bureau of Labor Statistics' Local Area Unemployment Statistics program (unemployment rate), we constructed a merged cross-sectional dataset covering 387 metropolitan statistical areas.

Exploratory analysis revealed a strong positive Pearson correlation of 0.81 between median income and median home value, while the correlation between unemployment rate and home value was near zero (r = 0.015). A multiple linear regression model confirmed that median household income is a statistically significant positive predictor of median home prices even after controlling for unemployment. These findings are consistent with economic theory predicting that higher purchasing power increases housing demand and prices. However, because the analysis is cross-sectional and observational, we interpret these results as associations rather than causal effects and discuss limitations including aggregation bias, omitted variables (e.g., housing supply, zoning), and the use of nominal rather than cost-of-living-adjusted dollars.

## Research Question

Is median household income associated with median home prices across U.S. metropolitan areas after controlling for local unemployment rates in 2023?

Using metropolitan-level data, we analyze median home price as a function of median household income, with unemployment rate included as a control variable because local labor market conditions may influence both household income levels and housing demand. Areas with higher unemployment may experience reduced purchasing power and lower housing demand, which could independently affect home prices. Controlling for unemployment therefore helps isolate the association between income and housing prices from broader economic conditions. We emphasize this control because unemployment can confound the income–price relationship: in weaker labor markets both income and housing demand tend to fall, so without holding unemployment constant we could misattribute to income alone effects that partly reflect local job conditions. We apply correlation analysis and multiple linear regression to assess whether income remains a statistically significant predictor of housing prices after accounting for these factors. This is a statistical inference task focused on estimating relationships rather than prediction.


## Background and Prior Work

Housing prices vary widely across U.S. metropolitan areas, reflecting differences in underlying regional economic conditions rather than a single determinant. Economic theory predicts that higher household income increases housing demand and can raise home prices, but empirical evidence suggests this relationship is not always stable. A Federal Reserve study examining metropolitan housing markets finds that the long-run relationship between income and house prices is weaker and less consistent than commonly assumed, indicating that additional local factors may influence price variation (Federal Reserve, 2003). Labor market conditions are one such factor, as unemployment affects income stability, credit access, and housing demand, and metropolitan-level data show substantial variation in unemployment rates across regions (Bureau of Labor Statistics, n.d.). Other peer-reviewed research using metropolitan housing data further demonstrates that housing prices respond to regional economic conditions in complex and heterogeneous ways across locations and populations (Journal of Public Economics study). These findings highlight that income alone may not fully explain price differences and that failing to account for broader economic conditions could bias estimates of the income–price relationship. Motivated by this literature, our analysis examines the association between median household income and metropolitan home prices while controlling for unemployment to better isolate the relationship of interest.

## Hypothesis


Housing prices are expected to be positively associated with local income levels across U.S. metropolitan areas. Metropolitan areas with higher median household income generally have stronger purchasing power, which increases housing demand and may contribute to higher median home values. Therefore, we hypothesize that median household income will be a strong positive predictor of median home value.

Because housing markets are also influenced by broader economic conditions, unemployment rate is included as a control variable. Areas with higher unemployment may experience weaker labor markets, reduced purchasing power, and lower housing demand, which could independently influence housing prices. Controlling for unemployment allows the analysis to isolate the relationship between income and housing prices more clearly.

Formally, the null hypothesis states that median household income has no association with median home values after controlling for unemployment rates (β₁ = 0). The alternative hypothesis predicts a positive association between income and home values (β₁ > 0), meaning that metropolitan areas with higher incomes are expected to exhibit higher median home prices even when accounting for differences in local labor market conditions.

Additionally, unemployment rates are expected to have a negative relationship with housing prices, reflecting weaker economic conditions and reduced housing demand in areas with higher unemployment.

### Multiple Linear Regression: Home Value ~ Income + Unemployment

To formally test our hypothesis we fit a multiple linear regression model:

**Median Home Value = β₀ + β₁ × Median Income + β₂ × Unemployment Rate + ε**

This allows us to estimate the association between income and home prices **while holding unemployment constant**. We use the `statsmodels` OLS implementation, which provides coefficient estimates, standard errors, p-values, and model diagnostics.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

from sklearn.linear_model import LinearRegression
from scipy import stats

# ── Multiple linear regression ──────────────────────────────────────────────
X = df[['median_income', 'unemployment_rate_2023']].copy()
y = df['median_home_value'].copy()

model = LinearRegression()
model.fit(X, y)

# coefficients
intercept = model.intercept_
coef_income = model.coef_[0]
coef_unemp = model.coef_[1]

# fitted values and residuals
fitted = model.predict(X)
residuals = y - fitted

# R-squared and adjusted R-squared
n = len(y)
p = X.shape[1]
r2 = model.score(X, y)
adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)

# ── Compute p-values manually ───────────────────────────────────────────────
X_with_const = np.column_stack([np.ones(n), X.values])

# residual variance estimate
rss = np.sum(residuals ** 2)
sigma2 = rss / (n - p - 1)

# variance-covariance matrix of coefficients
XtX_inv = np.linalg.inv(X_with_const.T @ X_with_const)
var_beta = sigma2 * XtX_inv
se_beta = np.sqrt(np.diag(var_beta))

# coefficients including intercept
beta = np.array([intercept, coef_income, coef_unemp])

# t-stats and p-values
t_stats = beta / se_beta
p_values = 2 * (1 - stats.t.cdf(np.abs(t_stats), df=n - p - 1))

# ── Print regression results ────────────────────────────────────────────────
print("Multiple Linear Regression Results")
print("=" * 50)
print(f"Intercept: {intercept:.4f} (p = {p_values[0]:.4e})")
print(f"Income coefficient: {coef_income:.4f} (p = {p_values[1]:.2e})")
print(f"Unemployment coefficient: {coef_unemp:.4f} (p = {p_values[2]:.4f})")
print(f"R-squared: {r2:.4f}")
print(f"Adj. R-squared: {adj_r2:.4f}")

# ── Residual diagnostics ────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Residuals vs Fitted
axes[0].scatter(fitted, residuals, alpha=0.4, s=15, color='steelblue', edgecolor='white', linewidth=0.3)
axes[0].axhline(0, color='red', ls='--', lw=1.5)
axes[0].set_xlabel('Fitted Values ($)')
axes[0].set_ylabel('Residuals ($)')
axes[0].set_title('Residuals vs. Fitted Values')
axes[0].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x/1000:.0f}K'))
axes[0].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x/1000:.0f}K'))

# Histogram of residuals
axes[1].hist(residuals, bins=30, color='coral', edgecolor='white')
axes[1].set_xlabel('Residual ($)')
axes[1].set_ylabel('Count')
axes[1].set_title('Distribution of Residuals')

# Q-Q plot
(osm, osr), (slope, intercept_qq, r) = stats.probplot(residuals, dist="norm")
axes[2].scatter(osm, osr, alpha=0.4, s=15, color='seagreen', edgecolor='white', linewidth=0.3)
axes[2].plot(osm, slope * np.array(osm) + intercept_qq, 'r--', lw=1.5)
axes[2].set_xlabel('Theoretical Quantiles')
axes[2].set_ylabel('Sample Quantiles')
axes[2].set_title('Q–Q Plot of Residuals')

plt.tight_layout()
plt.savefig('results/regression_diagnostics.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\nR-squared: {r2:.4f}")
print(f"Adj. R-squared: {adj_r2:.4f}")
print(f"Income coefficient: {coef_income:.4f} (p = {p_values[1]:.2e})")
print(f"Unemployment coefficient: {coef_unemp:.2f} (p = {p_values[2]:.4f})")
```

## Ethics

When examining the association between median household income and median home prices across U.S. metropolitan areas, several ethical considerations must be addressed.

**Aggregation and equity.** The study relies on aggregated metropolitan-level data, which risks obscuring important within-area inequalities — such as disparities by race, neighborhood, or socioeconomic status. A metro area can have a high median income while still containing communities experiencing poverty and housing insecurity. Acknowledging this limitation is essential to avoid overgeneralized conclusions that mask structural inequities.

**Normative framing of housing prices.** Interpreting the relationship between income and home prices requires caution to prevent normative assumptions that rising prices are universally beneficial. Higher housing costs can exacerbate affordability challenges and displacement for lower-income residents, even in metro areas where aggregate incomes are growing. Policy conclusions drawn from this analysis should not imply that income growth alone improves housing outcomes for all residents.

**Observational design and causal claims.** Because the analysis is observational and cross-sectional, it is ethically important to avoid policy recommendations that imply deterministic effects of income growth on housing markets without accounting for regulatory, historical, and structural factors (e.g., exclusionary zoning, redlining legacies, interest rate policy). Presenting associations as causal could mislead policymakers and the public.

**Transparency and reproducibility.** We use publicly available government data (Census ACS, BLS LAUS), clearly document all variable definitions, data cleaning steps, and modeling choices, and make our code available for inspection. This transparency is necessary to ensure reproducibility and to prevent misinterpretation of results that could influence housing or urban policy debates.


## Discussion and Conclusion

This study asked whether median household income is associated with median home prices across U.S. metropolitan areas after controlling for local unemployment rates. Using 2023 data from the Census Bureau (ACS) and the Bureau of Labor Statistics (LAUS) covering 387 metro areas, we found strong evidence supporting an affirmative answer.

**Summary of results.** Exploratory analysis revealed a Pearson correlation of **r = 0.805** between median income and median home value — by far the strongest bivariate relationship in our dataset. The unemployment rate showed virtually no linear association with home values on its own (r = 0.015), while its correlation with income was moderate and negative (r = −0.167). Our multiple linear regression model (Home Value ~ Income + Unemployment) confirmed that income is a highly statistically significant positive predictor of home prices even after controlling for unemployment (p < 0.001). The unemployment coefficient carried the expected negative sign, though its effect was substantially weaker than that of income. The model's R² indicates that income and unemployment together explain a meaningful share of cross-metro variation in home prices, though considerable residual variation remains.

**Connection to prior work.** These findings align with standard economic theory: higher incomes increase purchasing power and housing demand, putting upward pressure on prices in a relatively fixed housing supply. However, the Federal Reserve study we cited (Clark & Coggin, 2003) cautioned that the income–house-price relationship is not always stable over time or across market conditions. Our cross-sectional 2023 snapshot captures a single point in time and thus cannot address temporal instability. Still, the strong correlation we observe is consistent with the central prediction of the demand-side theory of housing prices.

**Interpreting the unemployment variable.** The near-zero bivariate correlation between unemployment and home value was somewhat surprising, given that economic theory predicts a negative relationship. One explanation is that unemployment rates were compressed across metros in 2023 due to a historically tight labor market (most metros clustered between 2–5%), leaving little variation to drive price differences. In the regression model, after controlling for income, unemployment did show a modest negative association with home prices, suggesting it captures some residual variation in local economic conditions beyond what income alone reflects.

**Limitations.** Several important limitations constrain interpretation. First, this is a **cross-sectional observational study**: we can identify associations but cannot make causal claims. Income may drive prices, prices may attract high-income residents, or both may respond to unobserved third variables (e.g., industry composition, amenities, housing supply constraints, zoning regulations). Second, **aggregation bias** is a concern: metro-level medians obscure within-area heterogeneity. A metro with high median income can still contain neighborhoods with severe affordability problems. Third, we use **nominal dollars** without adjusting for regional cost of living, so a given income level implies different purchasing power in different metros. Fourth, median home value covers only **owner-occupied housing** and excludes the rental market, which is especially relevant in metros with high renter populations. Fifth, with only **two predictors**, our model leaves substantial variance unexplained; supply-side factors (new construction permits, land availability, regulatory environment) are likely important omitted variables.

**Future directions.** Several extensions could strengthen the analysis. Incorporating **longitudinal data** (multiple years) would allow fixed-effects or difference-in-difference approaches to better control for unobserved metro-level confounders. Adding **supply-side variables** (e.g., housing permits, land-use regulation indices) and **demand-side controls** (e.g., mortgage rates, population growth) would provide a more complete picture. Examining **within-metro variation** using Census tract or ZIP-code data could address aggregation bias. Finally, adjusting incomes and home values for **regional price parity** (available from the Bureau of Economic Analysis) would improve the interpretability of cross-metro comparisons.

In conclusion, our analysis provides robust evidence that median household income is strongly and positively associated with median home values across U.S. metropolitan areas in 2023, and this association persists after controlling for local unemployment rates. These results support the hypothesis that purchasing power is a key factor underlying cross-metro housing price variation, while also highlighting the need for more comprehensive models that incorporate supply constraints and longitudinal dynamics to move from association toward causal understanding.
